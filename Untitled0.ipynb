```python
# Variables
name = "Ansh"
age = 20
pi = 3.14

print("Name:", name)
print("Age:", age)
print("Value of pi:", pi)
```

```python
# For loop
for i in range(5):   # runs from 0 to 4
    print("Iteration:", i)
```

```python
# While loop
count = 1
while count <= 5:
    print("Count:", count)
    count += 1
```

```python
# Function
def greet(name):
    return "Hello, " + name + "!"

print(greet("Ansh"))
print(greet("Python"))
```

```python
# List
fruits = ["apple", "banana", "cherry"]

print("First fruit:", fruits[0])   # indexing
fruits.append("orange")            # add item
print("Updated list:", fruits)

# Loop through list
for fruit in fruits:
    print("Fruit:", fruit)
```

```python
# Dictionary
student = {
    "name": "Ansh",
    "age": 20,
    "course": "Python Basics"
}

print("Name:", student["name"])
print("Age:", student["age"])

# Add new key-value
student["grade"] = "A"
print("Updated dictionary:", student)

# Loop through dictionary
for key, value in student.items():
    print(key, ":", value)
```

```python
# Mini program
students = [
    {"name": "Ansh", "age": 20},
    {"name": "Riya", "age": 22},
    {"name": "Aman", "age": 19}
]

def show_students(data):
    for student in data:
        print(student["name"], "is", student["age"], "years old.")

show_students(students)
```
